# [Recursion Guideline](https://inventwithpython.com/blog/leap-of-faith.html)

```python
def factorial(number):
    if number == 0 or number == 1:
        # BASE CASE
        return 1
    else:
        # RECURSIVE CASE
        return number * factorial(number - 1)
```

```python
factorial(5)  # Example usage, should return 120
```

The Five Guidelines for Taking a Leap of Faith

    - Start by figuring out the data types of the parameters and return value.
    - Next, implement the base case.
    - Take a leap of faith and assume your recursive function magically returns the correct value, and write your recursive case.
    - First Caveat: The argument to the recursive function call cannot be the original argument.
    - Second Caveat: The argument to the recursive function call must ALWAYS get closer to the base case.


```python
def permutations(chars):
    if len(chars) == 0 or len(chars) == 1:
        # BASE CASE
        return [chars]
    else:
        # RECURSIVE CASE
        all_perms = []  # Start with an empty list.
        head = chars[0]
        tail = chars[1:]

        tail_perms = permutations(tail)
        print(f"Tail permutations {tail_perms}")  # Debugging output
        for perm in tail_perms:  # Go through each of the tail's permutations.
            for i in range(len(perm) + 1):  # Go through each position and inject the head.
                new_perm = perm[:i] + head + perm[i:]
                all_perms.append(new_perm)  # Add it to the complete list of permutations.
        return all_perms  # Return all of the permutations.
```

```python
permutations("abcd")
```

```python
def permutations(chars):
    # BASE CASE: If the string is empty or has one letter, there's only one permutation.
    if len(chars) <= 1:
        return [chars]

    all_perms = []
    # RECURSIVE CASE:
    # For each character in the string...
    for i in range(len(chars)):
        # 1. Pick a character to be the "head".
        head = chars[i]

        # 2. Form the "rest" of the string.
        rest = chars[:i] + chars[i+1:] # Everything but the head

        # 3. Recursively find all permutations of the "rest".
        #    e.g., if head='B', rest='AC', this returns ['AC', 'CA']
        perms_of_rest = permutations(rest)

        # 4. Combine the "head" with each result from the recursive call.
        for p in perms_of_rest:
            all_perms.append(head + p)

    return all_perms

# Let's trace it for 'ABC'
# - When head='A', rest='BC'. permutations('BC') returns ['BC', 'CB']. We create ['ABC', 'ACB'].
# - When head='B', rest='AC'. permutations('AC') returns ['AC', 'CA']. We create ['BAC', 'BCA'].
# - When head='C', rest='AB'. permutations('AB') returns ['AB', 'BA']. We create ['CAB', 'CBA'].
```
